```python

# Connect to the webcam (0 = internal camera)
cam = cv2.VideoCapture(0)

# Click the photo
status, photo = cam.read()

# Save it
cv2.imwrite("my_photo.jpg", photo)

# Display the image
cv2.imshow("Captured Photo", photo)
cv2.waitKey(0) # Wait for any key to be pressed. 0 means wait indefinitely.
cv2.destroyAllWindows() # Close all OpenCV windows gracefully after the key press
```

```python
cam.release()
```

```python
cropped = photo[100:300, 200:400]  # rows 100-300, cols 200-400
cv2.imshow("Cropped", cropped)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
# Set the Blue channel to 0 for rows 20 to 40 → removing all blue in this region
photo[20:40, :, 0] = 0  # Blue 0
photo[20:40, :, 1] = 0  # Green 0
photo[20:40, :, 2] = 255  # Red 255 (max value)

# (from row 20 to 40) will appear in solid red color.

cv2.imshow("Red Highlight", photo)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
