```python
import cv2 
# Connect to the webcam (0 = internal camera)
cam = cv2.VideoCapture(0)

# Click the photo
status, photo = cam.read()

# Save it
cv2.imwrite("my_photo.jpg", photo)
cap = cv2.imread("my_photo.jpg")

# Display the image
cv2.imshow("Captured Photo", photo)
#cv2.waitKey() # Wait for any key to be pressed. 0 means wait indefinitely.
if cv2.waitKey() == 27 : #27 for esc
    cv2.destroyAllWindows() # Close all OpenCV windows gracefully after the key press
```

```python
cropped = photo[100:300, 200:400]  # rows 100-300, cols 200-400
cv2.imshow("Cropped", cropped)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
# Set the Blue channel to 0 for rows 20 to 40 → removing all blue in this region
photo[20:40, :, 0] = 0  # Blue 0
photo[20:40, :, 1] = 0  # Green 0
photo[20:40, :, 2] = 255  # Red 255 (max value)

# (from row 20 to 40) will appear in solid red color.

cv2.imshow("Red Highlight", photo)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
cap = cv2.VideoCapture(0)
status, photo = cap.read()

# Assuming you have loaded your image into 'img' variable
# cap = cv2.imread('.jpg')

# Convert to grayscale
gray_img = cv2.cvtColor(photo, cv2.COLOR_BGR2GRAY)

# Display or save the grayscale image
cv2.imshow('Grayscale Image', gray_img)
cv2.waitKey(0)
cv2.destroyAllWindows()

# If you want to save the grayscale image
cv2.imwrite('team13.png', gray_img)
cap.release()
```
